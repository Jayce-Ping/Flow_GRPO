```python
import os
import json
import shutil
```

```python
full_size_dir = 'half_2by2'
train_metadata_file = os.path.join(full_size_dir, 'train_metadata.jsonl')
test_metadata_file = os.path.join(full_size_dir, 'test_metadata.jsonl')
with open(train_metadata_file, 'r') as f:
    train_metadata = [json.loads(line) for line in f]

output_dir = 'half_2by2_micro_train'
output_height = 256
output_width = 256
os.makedirs(output_dir, exist_ok=True)
for item in train_metadata:
    item['height'] = output_height
    item['width'] = output_width

train_output_file = os.path.join(output_dir, 'train_metadata.jsonl')
with open(train_output_file, 'w') as f:
    for item in train_metadata:
        f.write(json.dumps(item) + '\n')

# Copy test metadata file
shutil.copy(test_metadata_file, os.path.join(output_dir, 'test_metadata.jsonl'))
```
