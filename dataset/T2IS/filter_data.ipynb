```python
import os
import json
data_summary_path = 'T2IS_data_with_editing_prompts.jsonl'
with open(data_summary_path, 'r') as f:
    data_summary = [json.loads(line) for line in f.readlines()]

data_summary = {item['idx']: item for item in data_summary}
```

```python
# Load existing train/test split
dataset_dir = 'train_all_2by2'
with open(os.path.join(dataset_dir, 'train_metadata.jsonl'), 'r') as f:
    trainset = [json.loads(line) for line in f.readlines()]
with open(os.path.join(dataset_dir, 'test_metadata.jsonl'), 'r') as f:
    testset = [json.loads(line) for line in f.readlines()]

# Update train/test with new data summary
for item in trainset:
    idx = item['idx']
    new_item = data_summary[idx]
    item.update(new_item)

for item in testset:
    idx = item['idx']
    new_item = data_summary[idx]
    item.update(new_item)

trainset = sorted(trainset, key=lambda x: x['idx'])
testset = sorted(testset, key=lambda x: x['idx'])

with open(os.path.join(dataset_dir, 'train_metadata.jsonl'), 'w') as f:
    for item in trainset:
        f.write(json.dumps(item) + '\n')
with open(os.path.join(dataset_dir, 'test_metadata.jsonl'), 'w') as f:
    for item in testset:
        f.write(json.dumps(item) + '\n')
```
